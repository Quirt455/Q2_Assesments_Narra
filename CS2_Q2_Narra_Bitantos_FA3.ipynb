```python
#Values
Iroman1 = 85
The_Hulk = 70
Spiderman1 = 90

#The list
scores = [Iroman1,The_Hulk,Spiderman1]

#For the display updated scores function if it stays false it knows you did not change a score
ScoreChange = False

#Prints all the choices you can pick
print("=== Movie Rating Analyzer ===")
print("Select the number of the operation you want to perform:")
print("1. Display Scores")
print("2. Search Score")
print("3. Compute Average Score")
print("4. Update Score")
print("5. Display Updated Scores")
print("6. Exit")


#Function to display score
def Display_Score():
  print("=== Movie Ratings Analyzer ===")
  print("Ironman 1:", scores[0])
  print("The Hulk:", scores[1])
  print("Spiderman 1:", scores[2])

#Function to search for the item that has the score you want to find
def Search_Scores():
  search = int(input("Enter score to search: "))

  for i in range(len(scores)):
    if scores[i] == search:
      print("Score found at index:", i)

    else:
      print("Score not found")

#Function that lets you print the average of all three movies rating
def Score_Average():
  print("=== Score Average ===")

  Average = (scores[0] + scores[1] + scores[2]) / 3
  print(Average)

#Function that lets you update the scores
def Update_Score():
  picked = int(input("what score do you want to change? (1-3) "))

  if picked == 1:
    scores[0] = int(input("Enter new score for Iroman 1: "))

  elif picked == 2:
    scores[1] = int(input("Enter new score for The hulk: "))

  elif picked == 3:
    scores[2] = int(input("Enter new score for Spiderman 1:"))

  else:
    print("Invalid choice")

#Lets you display the updated scores
def Display_Upd_Score():
  print("=== Updated Movie Ratings ===")
  print("Ironman 1:", scores[0])
  print("The Hulk:", scores[1])
  print("Spiderman 1:", scores[2])

#The function calls and loop
while True:
  choice = int(input("Enter your choice: "))

  if choice == 1:
    Display_Score()

  elif choice == 2:
    Search_Scores()

  elif choice == 3:
    Score_Average()

  elif choice == 4:
    Update_Score()
    ScoreChange = True

  elif choice == 5:
    if ScoreChange == True:
      Display_Upd_Score()

    elif ScoreChange == False:
      print("You hav'nt even changed any scores")

  elif choice == 6:
    print("Thanks for using this program")
    break

  else:
    print("Invalid choice")














```
